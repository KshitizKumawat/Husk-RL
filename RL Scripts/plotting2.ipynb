```python
import pandas as pd
```

```python
df = pd.read_csv('metric.csv')
df = df[df['dg_power'].isnull()]
```

```python
df
```

```python
import matplotlib.pyplot as plt

# Assuming df has columns: 'episode' and 'q_value'
plt.figure(figsize=(10, 5))
plt.plot(df['episode'], df['solar_power'], marker='o', linestyle='-')

plt.title('Episode vs Q-Value')
plt.xlabel('Episode')
plt.ylabel('Q-Value')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Assuming df has columns: 'episode' and 'q_value'
plt.figure(figsize=(10, 5))
plt.plot(df['episode'], df['load_power'], marker='o', linestyle='-')

plt.title('Episode vs Total Reward')
plt.xlabel('Episode')
plt.ylabel('Total Reward')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
fig, ax1 = plt.subplots(figsize=(10, 5))

# First y-axis for total_reward
ax1.set_xlabel('Episode')
ax1.set_ylabel('Total Reward', color='tab:blue')
ax1.plot(df['episode'], df['load_power'], color='tab:blue', label='Total Reward')
ax1.tick_params(axis='y', labelcolor='tab:blue')

# Second y-axis for q_value
ax2 = ax1.twinx()
ax2.set_ylabel('Q-Value', color='tab:orange')
ax2.plot(df['episode'], df['solar_power'], color='tab:orange', label='Q-Value')
ax2.tick_params(axis='y', labelcolor='tab:orange')

# Title and layout
plt.title('Episode vs Total Reward and Q-Value')
fig.tight_layout()
plt.show()
```
