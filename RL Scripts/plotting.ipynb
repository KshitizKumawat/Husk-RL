```python
import pandas as pd
import matplotlib.pyplot as plt
import time

# User input for the episode number
# episode_number = int(input("Enter the episode number to monitor: "))
plant_code = input("Enter the Plant Code:")
eps = int(input("Enter the episode number:"))

# plt.ion()  # Enable interactive mode
fig, ax = plt.subplots(figsize=(15, 5))

# Initialize empty line objects (they will be updated instead of redrawing the entire plot)
line1, = ax.plot([], [], label="Solar Power", marker="o")
line2, = ax.plot([], [], label="Load Power", marker="s")
line3, = ax.plot([], [], label="DG Power", marker=".")
line4, = ax.plot([], [], label="SOC", marker="^")
line5, = ax.plot([], [], label="Avg SOC", marker="d")
line6, = ax.plot([], [], label="Actual DG Power", marker="*")

plt.figure()
ax.set_xlabel("Time")
ax.set_ylabel("Values")
ax.set_title(f"Live Metrics for Episode {eps}, {plant_code}")
ax.legend()
ax.grid(True)
plt.xticks(rotation=45)

mng = plt.get_current_fig_manager()
try:
    mng.window.showMaximized()  # Works for Qt backend
except AttributeError:
    pass

def plot(plant_code,eps,trials,pause):
    t = trials
    while True:
    
        while True:
            try:
                df = pd.read_csv("metric.csv")
                break
            except:
                time.sleep(2)
                pass
            
        # Convert 'time' column to datetime
        df['time'] = pd.to_datetime(df['time'])
    
        # Filter data for the given episode
        # df_filtered = df[df['episode'] == episode_number]
        df_filtered = df[(df['pc'] == plant_code) & (df['episode']==eps)]
        
        # Check if there is data available for this episode
        if df_filtered.empty:
            print(f"No data available for episode: {eps} and PlantCode: {plant_code}. Retrying...")
            time.sleep(5)
            continue  # Skip iteration if no data
    
        # Update the data of each line (instead of clearing the entire plot)
        line1.set_xdata(df_filtered['time'])
        line1.set_ydata(df_filtered['solar_power'])
    
        line2.set_xdata(df_filtered['time'])
        line2.set_ydata(df_filtered['load_power'])
    
        line3.set_xdata(df_filtered['time'])
        line3.set_ydata(df_filtered['dg_power'])
    
        line4.set_xdata(df_filtered['time'])
        line4.set_ydata(df_filtered['soc'])
    
        line5.set_xdata(df_filtered['time'])
        line5.set_ydata(df_filtered['avg_soc'])
    
        line6.set_xdata(df_filtered['time'])
        line6.set_ydata(df_filtered['actual_dg_power'])
    
        # Update the x and y axis limits dynamically
        ax.set_xlim(df_filtered['time'].min(), df_filtered['time'].max())
        ax.set_ylim(
            min(df_filtered[['solar_power', 'load_power','dg_power' ,'soc', 'avg_soc']].min()), 
            110
        )
        ax.set_title(f"Live Metrics for Episode {eps}, {plant_code}")
        
        # Redraw the updated plot
        plt.draw()
        plt.show()
        plt.pause(pause)  # Wait 5 seconds before updating again
        
        if t==1:
            eps+=1
            t=trials
        else:
            t-=1
            
        print(eps)
    # break
    # plant_code = input("Enter the Plant Code:")

plot(plant_code,eps,trials=50,pause=13)
```
